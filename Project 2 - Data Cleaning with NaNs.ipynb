```python
import pandas as pd
import numpy as np

# new things: fillna(), dropna(), .groupby(), .agg()
```

```python
data = {
    'OrderID': [101, 102, 103, 104, 105, 106, 107, 108],
    'Category': ['Electronics', 'Clothing', 'Electronics', 'Home', 'Clothing', np.nan, 'Home', 'Electronics'],
    'Price': [299.99, np.nan, 150.00, 45.50, 25.00, 89.99, np.nan, 800.00],
    'Quantity': [1, 2, np.nan, 4, 1, 2, 1, 1]
}
df = pd.DataFrame(data)
```

```python
print(df)
```

```python
# .isna().sum()
print(df[["Category", "Price", "Quantity"]].isna().sum())
df = df.fillna({"Quantity": 1})
df = df.fillna({"Price": df["Price"].median()})
df = df.dropna()
df["Total_Sales"] = df["Price"] * df["Quantity"]
print(df)
print()
print()
print(df.groupby(["Category"]).agg({"Price": "mean", "Total_Sales": "sum"}))
```
