```python
import pandas as pd
import numpy as np

# Create a mock TSV trigger/event log with messy channel strings
mock_tsv_content = """trial_idx\tcue_label\treaction_time_ms\tbad_trial\tchannel_raw
0\tleft_hand\t450.2\tFalse\tCz.. 
1\tright_hand\t512.8\tFalse\t C3.
2\tleft_hand\tnan\tTrue\tC4.. 
3\tright_hand\t489.1\tFalse\tFz 
4\tleft_hand\t530.4\tFalse\t Cz 
5\tright_hand\t610.0\tTrue\tC3..
6\tleft_hand\t420.5\tFalse\tC4 
7\tright_hand\t495.3\tFalse\tCz..
"""

with open("experiment_events.tsv", "w") as f:
    f.write(mock_tsv_content)
```

```python
df = pd.read_csv("experiment_events.tsv", sep=r'\s+')
df["channel_raw"] = df["channel_raw"].str.strip(' .')
df = df[df["bad_trial"] == False]
hand_counts = df["cue_label"].value_counts()
correspondance = {
    "left_hand" : 0,
    "right_hand" : 1
}
df.insert(2, "target", df["cue_label"].map(correspondance))
reaction_time_means = df.groupby("cue_label").agg({"reaction_time_ms": "mean"})
y_labels = df["target"].to_numpy()

print(df)
print()
print(reaction_time_means)
print()
print(y_labels)
```
