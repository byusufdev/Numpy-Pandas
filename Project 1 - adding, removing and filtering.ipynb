```python
import numpy as np
import pandas as pd

# new things: pd.DataFrame(index= , columns =), .rename_axis(), .assign(), .insert(), .index.name(), .sum()
```

```python
daily_coffee_sales = np.random.randint(50, 150, size =(7,1))
df = pd.DataFrame(daily_coffee_sales,
                  index= ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"],
                  columns= ["Coffees_Sold"]).rename_axis("Day").assign(**{"Price_per_Coffee": 3.5})
# df.index.name = "Day"
# df.insert(1, "Price/Coffee", 3.5)
df.insert(2, "Daily_revenue", df["Coffees_Sold"]*df["Price_per_Coffee"])
print(df)
```

```python
df.at["Saturday", "Price_per_Coffee"] = 3.0
df.at["Sunday", "Price_per_Coffee"] = 3.0
# df.loc[["Saturday", "Sunday"], "Price_per_Coffee"] = 3.0

df["Daily_revenue"] = df["Coffees_Sold"] * df["Price_per_Coffee"]
print(df)
```

```python
print(np.mean(daily_coffee_sales))
# print(df["Coffees_Sold"].mean())
print(df["Daily_revenue"].sum(axis=0))
print(df[df["Daily_revenue"] > 300])
```
